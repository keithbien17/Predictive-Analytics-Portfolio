### Problem 2. Cleaning data and dealing with categorical features (Total Marks: 40)


**Q1.** 

- Remove all rows where `payment_default` contains missing values and reset the index (2 marks)  
- Using an appropriate `pandas` function, impute the remaining missing values with either the **mean** or the **mode**  (10 marks)  
  - Choose the imputation method based on the type of each variable and the best practices discussed in class and in the lecture notes while limiting your strategies to mean and mode 
- Explain what data imputation is, and describe how and why you applied it in this dataset  (3 marks)  

*(15 marks)*


```python
# Remove all rows where payment_default contains missing values 
df = df.dropna(subset=['payment_default'])

# reset the index.
df = df.reset_index(drop=True)

# Create a copy of the DataFrame to avoid SettingWithCopyWarning
df = df.copy()
```

```python
# Identify categorical/ordinal columns
categorical_cols = ['SEX', 'MARRIAGE', 'EDUCATION',
                    'PAY_1','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6']

# Identify numeric columns
numeric_cols = df.select_dtypes(include='number').columns
numeric_cols = [col for col in numeric_cols if col not in categorical_cols + ['payment_default']]

# Impute numeric columns with mean
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Impute categorical/ordinal columns with mode
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# check the applied imputation, missing values are fixed already
df.isnull().sum()
```

```python
df.head()
```

Ans: Data imputation is the process of filling in missing values in a dataset. This is important because it helps ensure the dataset is complete and suitable for analysis without losing data.In this dataset, rows with missing values in the target variable (payment_default) were removed because they cannot be used for prediction. Mean imputation was used for numerical variables, while mode imputation was used for categorical or nominal variables. It is needed in this dataset to handle missing values, reduce data loss, and maintain the overall structure and quality of the data for further analysis.If missing values are not handled, the dataset may become incomplete and can lead to inaccurate results or errors in analysis.

**Q4.**

- Encode the `MARRIAGE` feature using dummy variables, naming the new columns `MARRIAGE_MARRIED`, `MARRIAGE_SINGLE`, and `MARRIAGE_OTHER`, and add them to `df`  (5 marks)  
- Carefully determine how all values of `MARRIAGE` are mapped into these three dummy variables  (5 marks)  
  - Clearly explain any decisions or assumptions made
  - Remove the original `MARRIAGE` column from `df`  


(10 marks)   

```python
# Handle invalid values (0 → treat as OTHER)
df['MARRIAGE'] = df['MARRIAGE'].replace({0: 3})

# Create dummy variables
marriage_dummies = pd.get_dummies(df['MARRIAGE'])

# Rename columns
marriage_dummies = marriage_dummies.rename(columns={
    1: 'MARRIAGE_MARRIED',
    2: 'MARRIAGE_SINGLE',
    3: 'MARRIAGE_OTHER'
})

# Ensure values are 0/1 (not True/False)
marriage_dummies = marriage_dummies.astype(int)

# Add dummy variables to df
df = pd.concat([df, marriage_dummies], axis=1)

# Step 6: Remove original column
df = df.drop(columns=['MARRIAGE'])

# Step 7: Display output
df.head()
```

Ans: The MARRIAGE variable was encoded into three dummy variables: MARRIAGE_MARRIED, MARRIAGE_SINGLE, and MARRIAGE_OTHER. Each dummy variable represents a category, where a value of 1 indicates that the observation belongs to that category, and 0 otherwise.

The mapping was defined as follows: values equal to 1 were assigned to MARRIAGE_MARRIED = 1, and values equal to 2 were assigned to MARRIAGE_SINGLE = 1. All other values, including 0 and 3, were grouped into MARRIAGE_OTHER = 1.

This approach ensures that all observations are assigned to a category, including undefined or uncommon values.Finally, the original MARRIAGE column was removed from the dataset to avoid redundancy and multicollinearity.

### Problem 3: Preparing X and y arrays (Total Marks: 10)

**Q1.**

- Create a NumPy array `y` using the first 4,000 observations from the `payment_default` column of `df`  (2.5 marks)  
- Create a NumPy array `X` using the first 4,000 observations from all remaining variables in `df`  (2.5 marks)  

*(5 marks)*  

```python
# Crete y target variable

y = df['payment_default'].iloc[:4000].values
```

```python
# Create x features

X = df.drop(columns=['payment_default']).iloc[:4000].values
```

```python
# check shape
print(X.shape)
print(y.shape)
```

<hr style="width:25%;margin-left:0;"> 

**Q2.**

- Use an appropriate scikit-learn function covered in class to split the data into training and testing sets, creating `X_train`, `X_test`, `y_train`, and `y_test` with a 70%–30% split  (2.5 marks)  
  - Set `random_state = 777` and use stratification to ensure similar class proportions in both sets  
- Standardise the data using a scikit-learn function (2.5 marks)  

*(5 marks)*  


```python
#Train-test split (70–30 with stratification )

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=777,
    stratify=y
)
```

```python
# Standardise the data

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit on training data to avoid data leakage
X_train = scaler.fit_transform(X_train)

# Apply same transformation to test data
X_test = scaler.transform(X_test)
```

```python
print(X_train.shape, X_test.shape)
print(y_train.shape, y_test.shape)
```

### Problem 4: Generating Predictions and Choosing the Best Model (Total Marks: 30)


**Q1**. 

- Choose a linear classification model covered in the lectures and train it on the standardised data  (6 marks)  
- Compute and print the training and test set accuracies  (4 marks)

*(10 marks)*   

```python
# Linear model (Logistic Regression) 

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Train model
model_lr = LogisticRegression(max_iter=1000)
model_lr.fit(X_train, y_train)

# Predictions
y_train_pred = model_lr.predict(X_train)
y_test_pred = model_lr.predict(X_test)

# Accuracy
train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)


# Print results
print("Logistic Regression Train Accuracy:", train_acc)
print("Logistic Regression Test Accuracy:", test_acc)
```

**Q2**. 

- Choose a nonlinear classification model covered in the lectures and train it on the standardised data  (6 marks)   
- Compute and print the training and test set accuracies  (4 marks)   


*(10 marks)*  

```python
# Nonlinear model (Random Forest)

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Train model
model_rf = RandomForestClassifier(random_state=777)
model_rf.fit(X_train, y_train)

# Predictions
y_train_pred_rf = model_rf.predict(X_train)
y_test_pred_rf = model_rf.predict(X_test)

# Accuracy
train_acc_rf = accuracy_score(y_train, y_train_pred_rf)
test_acc_rf = accuracy_score(y_test, y_test_pred_rf)

# Print results
print("Random Forest Train Accuracy:", train_acc_rf)
print("Random Forest Test Accuracy:", test_acc_rf)
```
